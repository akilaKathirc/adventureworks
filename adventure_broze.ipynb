```python
input_path = "abfss://adventureworksbronze@dbrdollarstg.dfs.core.windows.net/"
input_parameter_path = "abfss://parameters@dbrdollarstg.dfs.core.windows.net/"
```

```python
df_parameters = spark.read.option("multiline", True).json(f"{input_parameter_path}/git.json")
display(df_parameters)
```

```python
dfs = {}

for row in df_parameters.collect():
    p_sink_folder = row['p_sink_folder']
    df_key = f"df_{p_sink_folder}"
    try:
        dfs[df_key] = spark.read.option("header", True).option("inferSchema", True)\
            .format("parquet")\
            .load(f"{input_path}/{p_sink_folder}/{p_sink_folder}.parquet")
        display(dfs[df_key])
    except:
        print(f"path doesn't exist for {df_key}")

# Example usage: access a DataFrame by its dynamic name
# display(dfs['df_some_folder'])
```
